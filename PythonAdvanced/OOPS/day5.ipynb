```python

from abc import ABC, abstractmethod


class Customer(ABC):
    @abstractmethod
    def get_discount(self):
        pass
    
class StandardCustomerDiscount(Customer):
    def get_discount(self):
        return  0.1
    
class NoCustomerDiscount(Customer):
    def get_discount(self):
        return  0

class DiscountManager:
    def calculate_discount(self,price:float, customer:Customer):
        return price * customer.get_discount()
    
dm = DiscountManager()
sd = StandardCustomerDiscount()

res = dm.calculate_discount(200,sd)
print(res)
```

```python
class Payment(ABC):
    @abstractmethod
    def get_method(self):
        pass
    
class CreditCard(Payment):
    def get_method(self):
        return "Credit Card"
    
class PayPal(Payment):
    def get_method(self):
        return "PayPal"

class PaymentProcessor:
    def process_payment(self, payment:Payment, amount:float):
        print(f"Processing {payment.get_method()} payment of ${amount}")
        
paypal = PayPal()

processor = PaymentProcessor()

processor.process_payment(paypal, 500)
       
```
