```python
import time

def fib(n):
    if n < 0:
        raise ValueError("Cant be negative")
    elif n < 2:
        return n
    else:
        return fib(n-1) + fib(n-2)
    
start = time.time()
for i in range(0, 35):
    print(fib(i),end=" ")
    
end = time.time()
print(f"\nTook {end - start:.4f} seconds")
```

```python
import time


def fib(n, cache=None):
    if cache is None:
        cache = {}
    
    if n in cache:
        return cache[n]
        
    if n < 0:
        raise ValueError("Can't be negative")
    elif n < 2:
        result = n
    else:
        result = fib(n-1, cache) + fib(n-2, cache)
    
    cache[n] = result
    return result

start = time.time()
for i in range(0,35):
    print(fib(i),end=" ")
    
end = time.time()
print(f"\nTook {end - start:.4f} seconds")
```
